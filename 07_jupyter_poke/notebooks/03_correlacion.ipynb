```python
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sys.path.append("..")

from scripts.database import engine

df = pd.read_sql("SELECT * FROM pokemon", engine)
```

```python
vars_num = ["height", "weight", "base_experience"]

corr = df[vars_num].corr()
corr
```

```python
plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Matriz de correlación Pokémon")
plt.show()
```

```python
sns.scatterplot(data=df, x="weight", y="base_experience")
plt.title("Peso vs Experiencia")
plt.show()
```

```python
import numpy as np
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18,5))

pares = [
    ('height', 'Altura'),
    ('weight', 'Peso'),
]

for ax, feature in zip(axes, ['height','weight','base_experience']):

    ax.scatter(df[feature], df['base_experience'], alpha=0.4)

    z = np.polyfit(df[feature], df['base_experience'], 1)
    p = np.poly1d(z)

    xp = np.linspace(df[feature].min(), df[feature].max(), 100)
    ax.plot(xp, p(xp), color='red')

    ax.set_title(feature)

plt.tight_layout()
plt.show()
```
